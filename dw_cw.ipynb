**Dataset Load**

```python
df = pd.read_csv('college_students_habits_1M.csv')
df.head()
```

**Dataset Info**

```python
#shape of the dataset
print("shape", df.shape)

#data type
print("datatype" , df.dtypes)

#summery of the dataset
df.describe()

```

**Check missing values**

```python
missing_values = df.isnull().sum()
print(missing_values)
```

**drop missing values**
(here no missing values)

```python
df = df.dropna()
```

**Remove duplicates**

```python
df = df.drop_duplicates()
print("After Remove duplicates" ,df.shape)
```

**Fix Column Name**

```python
#convert to the lower case
df.columns = df.columns.str.lower()

#replace space using _
df.columns = df.columns.str.replace(' ', '_')
df.head()

#replcace special characters
df.columns = df.columns.str.replace('[^a-zA-Z0-9_]', '', regex=True)

print("Cleaned column\n", df.columns)
```

**Convert datatypes**

```python
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].astype(str)
```

**Check outliers**

```python
import matplotlib.pyplot as plt

# Example: GPA distribution
df['gpa'].hist()
plt.title("GPA Distribution")
plt.show()
```

**Saving the cleared dataset**

```python
df.to_csv("Cleaned_student_dataset.csv", index=False)
print("Saved the cleaned dataset")
```
